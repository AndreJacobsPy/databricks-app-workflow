```python
from pyspark.sql import functions as F, types as T
from datetime import date

PERSON_SCHEMA = T.StructType(
    [
        T.StructField("person_id", T.IntegerType()),
        T.StructField("name", T.StringType()),
        T.StructField("birth_date", T.DateType()),
        T.StructField("occupation", T.StringType()),
        T.StructField("salary", T.DoubleType()),
    ]
)

# creating the dataframe for our dummy table
person_df = spark.createDataFrame(
    [
        (1, "John", date(1980, 1, 1), "Engineer", 70000),
        (2, "Andre", date(2000, 5, 14), "FDE", 100000),
    ],
    schema=PERSON_SCHEMA,
)
person_df.write.mode("overwrite").saveAsTable("bronze.person")
person_df.display()
```
